#### The missing values in this column are considered valid because projects with the statuses **'Not Started', 'On Hold', and 'Ongoing'** do not yet have an actual completion date. These projects have either not entered the execution phase or have not yet been completed. Therefore, the absence of an `Actual_End_Date` is expected and does not indicate a data-quality issue. Hence, these missing values are retained without applying any imputation techniques.
